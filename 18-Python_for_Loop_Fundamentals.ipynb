```python
# Write a program to print numbers from 1 to 100 using a for loop.

for i in list(range(1,101)):
    print(i)
```

```python
# Print all even numbers from 1 to 100.

for i in list(range(1,101)):
    if i%2==0:
        print(i)
```

```python
# Print all odd numbers from 1 to 100.
for i in list(range(1,101)):
    if i%2==0:
        continue
    else:
        print(i)
```

```python
# Take an integer n and print its multiplication table from 1 to 20.

number = int(input("Enter the input number:"))

for i in list(range(1,21)):
    print(f"{number} x {i} = {number*i}")
```

```python
# Calculate the sum of numbers from 1 to n using a loop.

n = int(input("Enter the value of n:"))
sum = 0

for i in list(range(1, n + 1)):
    sum = sum + i
print(sum)
```

```python
# Calculate the factorial of a number without using any built-in factorial function.

fact = 1
n = int(input("Enter the input number:"))

for i in list(range(1,n+1)):
    fact = fact * i
print(fact)
```

```python
# Print only the numbers divisible by 3

numbers = [12, 7, 9, 20, 33, 42, 8, 15]

for i in numbers:
    if i%3==0:
        print(i)
```

```python
# Print every language along with its length.

languages = ["Python", "Java", "C++", "JavaScript", "Go"]

for i in languages:
    print(f"Language is: {i} and length is: {len(i)}")
```

```python
# Iterate through and print every key and value.

student = {
			"name": "Rahul",
			"age": 22,
			"course": "Data Science",
			"city": "Bangalore"
          }

for key, value in student.items():
    print(f"Key is {key} and value is {value}")
```

```python
# Count how many vowels exist in a user-provided string.

string_value = input("Enter input string")
vowels = "aeiouAEIOU"
count = 0

for char in string_value:
    if char in vowels:
        count = count + 1
print(count)
```

```python
# Reverse a string using a for loop without using [::-1] or reversed().

string = input("Enter a string: ")
reversed_string = ""

for i in range(len(string) - 1, -1, -1):
    reversed_string += string[i]

print(f"Reversed: {reversed_string}")
```

```python
# Find the largest number from a list without using max().
l = [2,3,4,995,6,7,44,55,479]
n = l[0]

for i in l[1:]:
    if i > n:
        n = i
print(n)
```
