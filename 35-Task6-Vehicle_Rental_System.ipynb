<h3 style="font-size:16px; margin-bottom:4px;">Task 6: Vehicle Rental System</h3>

<p style="font-size:13px; margin:4px 0;">Build a mini vehicle rental application using inheritance: <code>Vehicle</code> is the parent class, with <code>Car</code> and <code>Bike</code> as child classes.</p>

<ul style="font-size:13px; margin:4px 0;">
<li><code>Vehicle</code> contains vehicle number, brand, model, and rental price per day.</li>
<li><code>Car</code> adds number of seats; <code>Bike</code> adds engine capacity.</li>
<li><code>calculate_rent(days)</code> computes the rental cost.</li>
<li>A static method validates that the rental duration (days) is greater than zero.</li>
<li>Create at least 2 cars and 2 bikes and demonstrate the complete system.</li>
</ul>

<h4 style="font-size:14px; margin-bottom:2px;">Parent class: Vehicle</h4>

```python
class Vehicle:
    """Base class representing a rentable vehicle."""

    def __init__(self, vehicle_number, brand, model, rental_price_per_day):
        """Initialize a vehicle with number, brand, model, and rental price per day."""
        self.vehicle_number = vehicle_number
        self.brand = brand
        self.model = model
        self.rental_price_per_day = rental_price_per_day

    def display_details(self):
        """Print a formatted summary of the vehicle's details."""
        print(f"Vehicle Number : {self.vehicle_number}")
        print(f"Brand          : {self.brand}")
        print(f"Model          : {self.model}")
        print(f"Rent Per Day   : {self.rental_price_per_day}")

    def calculate_rent(self, days):
        """Return the total rent for renting this vehicle for the given number of days."""
        return self.rental_price_per_day * days

    @staticmethod
    def is_valid_duration(days):
        """Return True if the rental duration is valid (greater than zero), otherwise False."""
        return days > 0
```

<h4 style="font-size:14px; margin-bottom:2px;">Child classes: Car and Bike (inherit from Vehicle)</h4>

```python
class Car(Vehicle):
    """Child class that extends Vehicle with number of seats."""

    def __init__(self, vehicle_number, brand, model, rental_price_per_day, seats):
        """Initialize a car using the parent Vehicle constructor, plus number of seats."""
        super().__init__(vehicle_number, brand, model, rental_price_per_day)
        self.seats = seats

    def display_details(self):
        """Print the vehicle details from the parent class, plus the number of seats."""
        super().display_details()
        print(f"Seats          : {self.seats}")


class Bike(Vehicle):
    """Child class that extends Vehicle with engine capacity."""

    def __init__(self, vehicle_number, brand, model, rental_price_per_day, engine_capacity):
        """Initialize a bike using the parent Vehicle constructor, plus engine capacity."""
        super().__init__(vehicle_number, brand, model, rental_price_per_day)
        self.engine_capacity = engine_capacity

    def display_details(self):
        """Print the vehicle details from the parent class, plus the engine capacity."""
        super().display_details()
        print(f"Engine Capacity: {self.engine_capacity}")
```

<h4 style="font-size:14px; margin-bottom:2px;">Create at least 2 cars and 2 bikes</h4>

```python
car1 = Car("KA01AB1234", "Toyota", "Innova", 2500, 7)
car2 = Car("KA02CD5678", "Hyundai", "i20", 1500, 5)
bike1 = Bike("KA03EF9012", "Royal Enfield", "Classic 350", 800, 350)
bike2 = Bike("KA04GH3456", "Honda", "Activa", 400, 110)

vehicles = [car1, car2, bike1, bike2]

for vehicle in vehicles:
    vehicle.display_details()
    print("-" * 30)
```

<h4 style="font-size:14px; margin-bottom:2px;">Calculate rent, including duration validation</h4>

```python
def rent_vehicle(vehicle, days):
    """Validate the rental duration, then print the total rent for a vehicle."""
    if not Vehicle.is_valid_duration(days):
        print(f"Invalid rental duration ({days} days) for {vehicle.brand} {vehicle.model}.")
        return
    total_rent = vehicle.calculate_rent(days)
    print(f"Renting {vehicle.brand} {vehicle.model} for {days} days costs: {total_rent}")

rent_vehicle(car1, 5)
rent_vehicle(bike1, 3)
rent_vehicle(car2, 2)
rent_vehicle(bike2, 7)

# Demonstrate the duration validation catching an invalid rental request.
rent_vehicle(car1, 0)
rent_vehicle(bike2, -4)
```
