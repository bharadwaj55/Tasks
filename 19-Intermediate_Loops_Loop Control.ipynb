```python
# Print numbers from 1–100 but skip numbers divisible by 5 using continue.

for i in list(range(1,101)):
    if i%5==0:
        continue
    print(i)
```

```python
# Iterate from 1–100 and stop when you encounter the first number divisible by both 7 and 11.
# Stops when first number found, Doesn't print 77 (breaks before printing)

for i in list(range(1,101)):
    if i%7==0 and i%11==0:
        break
    print(i)
```

```python
# Search for a user-provided number inside a list. Use for-else to print:
# "Number Found"
# "Number Not Found"

# The else block executes only if the loop completes without hitting break
# If number is found → print "Number Found" and break (else doesn't run)
# If number is not found → loop completes → else runs and prints "Number Not Found"


l = [1,2,3,4,5,6,7,8]
n = int(input("Enter input number"))

for i in l:
    if i==n:
        print("Number Found")
        break
else:
    print("Number Not Found")
```

```python
# Generate multiplication tables from 1 to 10 using nested loops.

for i in list(range(1,11)):
    for j in range(1,11):
        print(f"{i} x {j} = {i*j}")
    print()
```

```python
# Given a list containing duplicate elements, 
# create another list containing only unique elements without using set().

l = [10,10,10,20,30,50,50,70,80,90,90]
unique_list = []

for i in l:
    if i not in unique_list:
        unique_list.append(i)
print(unique_list)
```

```python
# Find # count, positive numbers, negative numbers, zeros

numbers = [10, -4, 8, -2, 0, 15, -9, 21]
print(f"Count of numbers is {len(numbers)}")
```

```python
# Find positive numbers

numbers = [10, -4, 8, -2, 0, 15, -9, 21]
positive_numbers = []

for i in numbers:
    if i > 0:
        positive_numbers.append(i)
print(positive_numbers)
```

```python
# Find negative numbers

numbers = [10, -4, 8, -2, 0, 15, -9, 21]
negative_numbers = []

for i in numbers:
    if i < 0:
        negative_numbers.append(i)
print(negative_numbers)
```

```python
# Find zeroes

numbers = [10, -4, 8, -2, 0, 15, -9, 21]
zero_list = []

for i in numbers:
    if i == 0:
        zero_list.append(i)
print(zero_list)
```

```python
# Write a program to determine whether a number is prime using a loop.
# A prime number is a natural number greater than 1 that has exactly two divisors: 1 and itself. 
# It cannot be divided evenly by any other number.

n = int(input("Enter a number: "))
is_prime = True

if n < 2:
    is_prime = False
else:
    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break

if is_prime:
    print(f"{n} is a prime number")
else:
    print(f"{n} is not a prime number")
```

```python
# Print all prime numbers between 1 and 100.

for n in range(2, 101):
    is_prime = True
    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break
    if is_prime:
        print(f"The number {n} is prime")

```
