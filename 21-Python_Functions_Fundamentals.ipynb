```python
# Create def add(a, b) that returns the addition of two numbers.

def add(a, b):
    """Return the sum of two numbers."""
    return a + b

print(add(5, 3))
```

```python
# Create functions for: addition, subtraction, multiplication, division

def addition(a, b):
    """Return the sum of two numbers."""
    return a + b

def subtraction(a, b):
    """Return the difference between two numbers."""
    return a - b

def multiplication(a, b):
    """Return the product of two numbers."""
    return a * b

def division(a, b):
    """Return the quotient of two numbers."""
    return a / b

print(f"Addition: {addition(10, 4)}")
print(f"Subtraction: {subtraction(10, 4)}")
print(f"Multiplication: {multiplication(10, 4)}")
print(f"Division: {division(10, 4)}")
```

```python
# Create a function that determines whether a number is even or odd.

def even_or_odd(number):
    """Return 'Even' or 'Odd' depending on the given number."""
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(even_or_odd(7))
print(even_or_odd(12))
```

```python
# Create a function that returns the largest of three numbers without using max().

def largest_of_three(a, b, c):
    """Return the largest of three numbers."""
    largest = a
    if b > largest:
        largest = b
    if c > largest:
        largest = c
    return largest

print(largest_of_three(15, 42, 8))
```

```python
# Create a function that calculates factorial.

def factorial(n):
    """Return the factorial of a non-negative integer n."""
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
print(factorial(0))
```

```python
# Create a function that checks whether a number is prime.

def is_prime(n):
    """Return True if n is a prime number, otherwise False."""
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(17))
print(is_prime(15))
```

```python
# Create def calculate_discount(price, discount=10) where the default discount is 10%.

def calculate_discount(price, discount=10):
    """Return the price after applying the given discount percentage."""
    discounted_price = price - (price * discount / 100)
    return discounted_price

print(calculate_discount(1000))
print(calculate_discount(1000, 20))
```

```python
# Create a function that accepts a list and returns its sum without using sum().

def list_sum(numbers):
    """Return the sum of all numbers in a list."""
    total = 0
    for number in numbers:
        total = total + number
    return total

print(list_sum([1, 2, 3, 4, 5]))
```

```python
# Create a function that accepts a string and returns the number of vowels.

def count_vowels(text):
    """Return the count of vowels in the given string."""
    vowels = "aeiouAEIOU"
    count = 0
    for char in text:
        if char in vowels:
            count = count + 1
    return count

print(count_vowels("Hello World"))
```

```python
# Create a function that accepts a string and determines whether it is a palindrome.

def is_palindrome(text):
    """Return True if the given string reads the same forwards and backwards."""
    reversed_text = text[::-1]
    return text == reversed_text

print(is_palindrome("madam"))
print(is_palindrome("python"))
```

```python
# Create a function that accepts name, age, course and returns a formatted student profile.
# Use both positional and keyword arguments while calling it.

def student_profile(name, age, course):
    """Return a formatted student profile string."""
    return f"Name: {name}, Age: {age}, Course: {course}"

# Positional arguments
print(student_profile("Riya", 21, "Data Science"))

# Keyword arguments
print(student_profile(name="Arjun", age=23, course="Python Development"))
```

```python
# Demonstrate the difference between a function that uses print and a function that uses return.

def print_square(n):
    """Print the square of a number but return nothing (returns None)."""
    print(n * n)

def return_square(n):
    """Return the square of a number so it can be stored and reused."""
    return n * n

# A print function only displays the value; it cannot be captured or reused.
print_result = print_square(4)
print(f"print_result holds: {print_result}")

# A return function gives back a value that can be stored and used in further calculations.
return_result = return_square(4)
print(f"return_result holds: {return_result}")

# Because return_square() gives back a real value, it can be reused in another calculation.
total = return_result + return_square(5)
print(f"Sum of two squared values: {total}")
```
