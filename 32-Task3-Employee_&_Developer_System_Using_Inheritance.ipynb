<ul style="font-size:13px; margin:4px 0;">
<li>Create a base class called <code>Employee</code> containing employee ID, name, salary, and department, along with a method <code>display_details()</code>.</li>
<li>Create a child class <code>Developer</code> that inherits from <code>Employee</code> and adds programming language and experience.</li>
<li>Create at least 3 Employee/Developer objects and demonstrate how the child class can access functionality from the parent class.</li>
</ul>

<p style="font-size:13px; margin:4px 0;">Concepts used: Parent class, child class, inheritance, object creation.</p>

<h4 style="font-size:14px; margin-bottom:2px;">Parent class: Employee</h4>

```python
class Employee:
    """Base class representing a general employee."""

    def __init__(self, employee_id, name, salary, department):
        """Initialize an employee with ID, name, salary, and department."""
        self.employee_id = employee_id
        self.name = name
        self.salary = salary
        self.department = department

    def display_details(self):
        """Print a formatted summary of the employee's details."""
        print(f"Employee ID : {self.employee_id}")
        print(f"Name        : {self.name}")
        print(f"Salary      : {self.salary}")
        print(f"Department  : {self.department}")
```

<h4 style="font-size:14px; margin-bottom:2px;">Child class: Developer (inherits from Employee)</h4>

```python
class Developer(Employee):
    """Child class that extends Employee with programming language and experience."""

    def __init__(self, employee_id, name, salary, department, programming_language, experience):
        """Initialize a developer using the parent Employee constructor, plus extra fields."""
        # Reuse the parent class's __init__ to set the shared employee fields.
        super().__init__(employee_id, name, salary, department)
        self.programming_language = programming_language
        self.experience = experience

    def display_details(self):
        """Print the employee details from the parent class, plus developer-specific details."""
        # Call the parent class's method to reuse its display logic.
        super().display_details()
        print(f"Language    : {self.programming_language}")
        print(f"Experience  : {self.experience} years")
```

<h4 style="font-size:14px; margin-bottom:2px;">Create Employee and Developer objects</h4>

```python
employee1 = Employee("E101", "Sunita", 55000, "Human Resources")
developer1 = Developer("E102", "Rahul", 85000, "Engineering", "Python", 4)
developer2 = Developer("E103", "Priya", 90000, "Engineering", "JavaScript", 6)

for person in (employee1, developer1, developer2):
    person.display_details()
    print("-" * 30)
```

<h4 style="font-size:14px; margin-bottom:2px;">Child class accessing parent class functionality</h4>

```python
# Developer did not redefine these attributes/behaviour itself; it inherited them from Employee.
print(f"{developer1.name} works in the '{developer1.department}' department (inherited attribute).")
print(f"{developer1.name}'s salary, set via Employee.__init__: {developer1.salary}")

# isinstance shows a Developer object is also an Employee, because of inheritance.
print(f"Is developer1 an instance of Employee? {isinstance(developer1, Employee)}")

# Developer.display_details() calls super().display_details(), reusing the parent's method.
print("Developer details (parent fields + child fields):")
developer2.display_details()
```
