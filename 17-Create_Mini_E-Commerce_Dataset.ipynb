```python
products = [
				{
				"name": "Laptop",
				"price": 70000,
				"brand": "Dell"
				},
				{
				"name": "Phone",
				"price": 40000,
				"brand": "Samsung"
				},
				{
				"name": "Tablet",
				"price": 30000,
				"brand": "Apple"
				}
           ]
```

```python
# Print all products
print(products)
```

```python
# Print first product
print(products[0])
```

```python
# Print second product's price
print(products[1]['price'])
```

```python
# Print third product's brand
print(products[2]['brand'])
```

```python
# Change first product's price
products[0]['price'] = "100000"
print(products)
```

```python
# Add "rating" to the second product
products[1]['rating'] = 5
print(products)
```

```python
# Add another product manually
products.append({'name': 'Ipad', 'price': '55000', 'brand': 'Apple'})
```

```python
# Print the final dataset
print(products)
```

## How lists, dictionaries, indexing, keys, and values work together

`products` is a **list**, and each element inside it is a **dictionary** representing one product.

- **Indexing** (`products[0]`, `products[1]`, ...) selects one product (one dictionary) from the list by its position.
- Each product dictionary stores its details as **key-value pairs**, e.g. `"name": "Laptop"`, `"price": 70000`, `"brand": "Dell"`.
- **Keys** (`"name"`, `"price"`, `"brand"`) are used to access or update the corresponding **values** inside a specific product, e.g. `products[1]['price']` first indexes into the list to get the second product, then uses the `'price'` key to get its value.

So indexing narrows the list down to a single dictionary, and keys then narrow that dictionary down to a single value.
