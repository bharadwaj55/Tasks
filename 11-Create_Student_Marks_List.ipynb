```python
marks = [78, 85, 90, 67, 88, 92, 76]
```

```python
# Print the complete list
print(marks)
```

```python
# Print the first element
print(marks[0])
```

```python
# Print the last element
print(marks[-1])
```

```python
# Print elements from index 2 to 5
print(marks[2:6])
```

```python
# Find the number of elements
print(len(marks))
```

```python
# Find maximum marks
print(max(marks))
```

```python
# Find minimum marks
print(min(marks))
```

```python
# Find total marks
print(sum(marks))
```

```python
# Sort marks in ascending order
# sorted() returns a new list and leaves the original 'marks' unchanged
print(sorted(marks))
```

```python
# Sort marks in descending order
# reverse=True sorts highest to lowest; 'marks' itself still stays unchanged
print(sorted(marks, reverse=True))
```

```python
# Add 95
marks.append(95)
print(marks)
```

```python
# Add [81, 84]
# extend() adds each item individually; 

marks.extend([81, 84])
print(marks)
```

```python
# append() would add [81, 84] as one nested element

new_marks = [78, 85, 90, 67, 88, 92, 76]
new_marks.append([81, 84])
print(new_marks)
```

```python
# Remove 67
# remove() raises ValueError if 67 isn't in the list
marks.remove(67)
print(marks)
```

```python
# Count how many times 90 occurs
print(marks.count(90))
```

```python
# Find the index of 88
print(marks.index(88))
```
