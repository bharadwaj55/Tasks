<h3 style="font-size:16px; margin-bottom:4px;">Task 7: Super30 Learning Platform</h3>

<p style="font-size:13px; margin:4px 0;">Build a mini Super30 Learning Platform using Python OOP.</p>

<ul style="font-size:13px; margin:4px 0;">
<li><code>User</code> (base class): name, email, user ID.</li>
<li><code>Student(User)</code>: course name and completed assignments.</li>
<li><code>Mentor(User)</code>: expertise and number of students assigned.</li>
</ul>

<p style="font-size:13px; margin:4px 0;">Concepts used: inheritance, class variable, class method, static method, constructors, instance methods, multiple objects.</p>

```python
class User:
    """Base class representing a user of the Super30 Learning Platform."""

    # Class variable shared by every instance, tracks how many users exist.
    total_users = 0

    def __init__(self, name, email, user_id):
        """Initialize a user with their basic details."""
        self.name = name
        self.email = email
        self.user_id = user_id

        User.total_users += 1

    def display_info(self):
        """Print the user's basic details."""
        print(f"User ID : {self.user_id}")
        print(f"Name    : {self.name}")
        print(f"Email   : {self.email}")

    @classmethod
    def get_total_users(cls):
        """Return the total number of User objects created so far (across all subclasses)."""
        return cls.total_users

    @staticmethod
    def is_valid_email(email):
        """Return True if email looks like a valid, simple email address."""
        return "@" in email and "." in email.split("@")[-1]
```

<h4 style="font-size:14px; margin-bottom:2px;">Student class (inherits from User)</h4>

```python
class Student(User):
    """Represents a student, extending User with course and assignment details."""

    def __init__(self, name, email, user_id, course_name=None):
        """Initialize a student, optionally with a course already assigned."""
        super().__init__(name, email, user_id)
        self.course_name = course_name
        self.completed_assignments = []

    def assign_course(self, course_name):
        """Assign or change the student's course."""
        self.course_name = course_name

    def submit_assignment(self, assignment_name):
        """Record a completed assignment for the student."""
        self.completed_assignments.append(assignment_name)

    def display_info(self):
        """Print the student's full details, including course and assignments."""
        super().display_info()
        print(f"Course  : {self.course_name}")
        print(f"Assignments Completed : {self.completed_assignments}")
```

<h4 style="font-size:14px; margin-bottom:2px;">Mentor class (inherits from User)</h4>

```python
class Mentor(User):
    """Represents a mentor, extending User with expertise and assigned students."""

    def __init__(self, name, email, user_id, expertise):
        """Initialize a mentor with their area of expertise."""
        super().__init__(name, email, user_id)
        self.expertise = expertise
        self.students_assigned = 0

    def add_student(self):
        """Increase the mentor's assigned student count by one."""
        self.students_assigned += 1

    def display_info(self):
        """Print the mentor's full details, including expertise and student count."""
        super().display_info()
        print(f"Expertise         : {self.expertise}")
        print(f"Students Assigned : {self.students_assigned}")
```

<h4 style="font-size:14px; margin-bottom:2px;">Create multiple objects</h4>

```python
student1 = Student("Riya", "riya@example.com", "U101")
student2 = Student("Arjun", "arjun@example.com", "U102", "Python Development")

mentor1 = Mentor("Dr. Verma", "verma@example.com", "U201", "Data Science")
```

<h4 style="font-size:14px; margin-bottom:2px;">Register a course, submit assignments, assign students to a mentor</h4>

```python
student1.assign_course("Data Science")
student1.submit_assignment("Assignment 1")
student1.submit_assignment("Assignment 2")

student2.submit_assignment("Assignment 1")

mentor1.add_student()
mentor1.add_student()

print(f"{student1.name} is now enrolled in {student1.course_name}")
print(f"{mentor1.name} now has {mentor1.students_assigned} students assigned")
```

<h4 style="font-size:14px; margin-bottom:2px;">Display student and mentor information</h4>

```python
for student in (student1, student2):
    student.display_info()
    print("-" * 30)

mentor1.display_info()
print("-" * 30)
```

<h4 style="font-size:14px; margin-bottom:2px;">Count total users and validate emails</h4>

```python
print(f"Total users on the platform: {User.get_total_users()}")
print(f"Same value accessed via a Student instance: {student1.get_total_users()}")
print(f"Is 'riya@example.com' a valid email? {User.is_valid_email('riya@example.com')}")
print(f"Is 'not-an-email' a valid email? {User.is_valid_email('not-an-email')}")
```
