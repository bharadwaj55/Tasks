<h3 style="font-size:16px; margin-bottom:4px;">Task 4: E-Commerce Product System</h3>

<ul style="font-size:13px; margin:4px 0;">
<li>Create a <code>Product</code> class with product ID, name, price, category, and stock quantity.</li>
<li>Implement <code>display_product()</code>, <code>update_stock()</code>, and <code>calculate_total_price(quantity)</code>.</li>
<li>Create a static method <code>Product.is_valid_price(price)</code> that returns <code>True</code> if the price is greater than zero, otherwise <code>False</code>.</li>
<li>Create at least 5 different products and demonstrate buying/updating stock.</li>
</ul>

```python
class Product:
    """Represents a single product in the e-commerce catalog."""

    def __init__(self, product_id, name, price, category, stock_quantity):
        """Initialize a product with ID, name, price, category, and stock quantity."""
        self.product_id = product_id
        self.name = name
        self.price = price
        self.category = category
        self.stock_quantity = stock_quantity

    def display_product(self):
        """Print a formatted summary of the product's details."""
        print(f"Product ID : {self.product_id}")
        print(f"Name       : {self.name}")
        print(f"Price      : {self.price}")
        print(f"Category   : {self.category}")
        print(f"Stock      : {self.stock_quantity}")

    def update_stock(self, change):
        """Adjust the stock quantity by change (positive to restock, negative to sell)."""
        if self.stock_quantity + change < 0:
            print(f"Cannot update stock: only {self.stock_quantity} units available.")
            return
        self.stock_quantity += change
        print(f"Stock updated for {self.name}. New stock: {self.stock_quantity}")

    def calculate_total_price(self, quantity):
        """Return the total price for buying the given quantity of this product."""
        return self.price * quantity

    @staticmethod
    def is_valid_price(price):
        """Return True if the given price is greater than zero, otherwise False."""
        return price > 0
```

<h4 style="font-size:14px; margin-bottom:2px;">Create at least 5 different products</h4>

```python
product1 = Product("P201", "Wireless Mouse", 799, "Electronics", 50)
product2 = Product("P202", "Running Shoes", 2499, "Footwear", 30)
product3 = Product("P203", "Coffee Mug", 249, "Kitchenware", 100)
product4 = Product("P204", "Notebook", 60, "Stationery", 200)
product5 = Product("P205", "Bluetooth Speaker", 1999, "Electronics", 25)

products = [product1, product2, product3, product4, product5]

for product in products:
    product.display_product()
    print("-" * 30)
```

<h4 style="font-size:14px; margin-bottom:2px;">Static method: check price validity</h4>

```python
print(f"Is 799 a valid price? {Product.is_valid_price(799)}")
print(f"Is -50 a valid price? {Product.is_valid_price(-50)}")
print(f"Is 0 a valid price? {Product.is_valid_price(0)}")
```

<h4 style="font-size:14px; margin-bottom:2px;">Demonstrate buying and updating stock</h4>

```python
# Buying 3 units of product1: reduce stock and calculate the total price.
quantity = 3
total_cost = product1.calculate_total_price(quantity)
print(f"Buying {quantity} x {product1.name} costs: {total_cost}")
product1.update_stock(-quantity)

# Restocking product3 with 20 more units.
product3.update_stock(20)

# Attempting to buy more units of product5 than are in stock.
product5.update_stock(-100)
```
