<h3 style="font-size:16px; margin-bottom:4px;">1. Student Result Management System</h3>

<p style="font-size:13px; margin:4px 0;">Create functions to accept student marks, calculate total, calculate percentage, assign grade, determine pass/fail, and display the result. Use loops wherever appropriate.</p>

```python
def accept_marks(num_subjects):
    """Ask the user to enter marks for each subject and return them as a list."""
    marks = []
    for subject_number in range(1, num_subjects + 1):
        mark = float(input(f"Enter marks for subject {subject_number} (out of 100): "))
        marks.append(mark)
    return marks


def calculate_total(marks):
    """Return the sum of all marks using a loop."""
    total = 0
    for mark in marks:
        total += mark
    return total


def calculate_percentage(total, num_subjects, max_marks_per_subject=100):
    """Return the percentage scored out of the maximum possible marks."""
    max_total = num_subjects * max_marks_per_subject
    return (total / max_total) * 100


def assign_grade(percentage):
    """Return a letter grade based on the percentage scored."""
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 35:
        return "D"
    else:
        return "F"


def determine_pass_fail(marks, passing_mark=35):
    """Return Pass only if every subject meets the passing mark, else Fail."""
    for mark in marks:
        if mark < passing_mark:
            return "Fail"
    return "Pass"


def display_result(student_name, marks, total, percentage, grade, status):
    """Print a formatted summary of the student's result."""
    print(f"Student Name : {student_name}")
    print(f"Marks        : {marks}")
    print(f"Total        : {total}")
    print(f"Percentage   : {percentage:.2f}%")
    print(f"Grade        : {grade}")
    print(f"Result       : {status}")
```

```python
student_name = input("Enter student name: ")
num_subjects = int(input("Enter number of subjects: "))

marks = accept_marks(num_subjects)
total = calculate_total(marks)
percentage = calculate_percentage(total, num_subjects)
grade = assign_grade(percentage)
status = determine_pass_fail(marks)

display_result(student_name, marks, total, percentage, grade, status)
```

```python
# Sample run with hardcoded data (no input() needed)

student_name = "Riya"
num_subjects = 3
marks = [85.0, 92.0, 78.0]

total = calculate_total(marks)
percentage = calculate_percentage(total, num_subjects)
grade = assign_grade(percentage)
status = determine_pass_fail(marks)

display_result(student_name, marks, total, percentage, grade, status)
```

<h3 style="font-size:16px; margin-bottom:4px;">2. Banking Application</h3>

<p style="font-size:13px; margin:4px 0;">A menu-driven banking program supporting check balance, deposit, withdraw, transaction history, and exit. Each operation is a function, and a while loop drives the menu.</p>

```python
def check_balance(balance):
    """Print the current account balance."""
    print(f"Current Balance: {balance:.2f}")


def deposit(balance, history, amount):
    """Add the amount to the balance and record the transaction."""
    balance += amount
    history.append(f"Deposited {amount:.2f}")
    print(f"Deposited {amount:.2f}. New balance: {balance:.2f}")
    return balance


def withdraw(balance, history, amount):
    """Subtract the amount from the balance if funds are sufficient."""
    if amount > balance:
        print("Insufficient balance.")
    else:
        balance -= amount
        history.append(f"Withdrew {amount:.2f}")
        print(f"Withdrew {amount:.2f}. New balance: {balance:.2f}")
    return balance


def show_transaction_history(history):
    """Print every transaction recorded so far."""
    if not history:
        print("No transactions yet.")
        return
    for index, transaction in enumerate(history, start=1):
        print(f"{index}. {transaction}")
```

```python
balance = 0.0
history = []

while True:
    print("\n----- Banking Menu -----")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Transaction History")
    print("5. Exit")
    choice = input("Enter your choice (1-5): ")

    if choice == "1":
        check_balance(balance)
    elif choice == "2":
        amount = float(input("Enter amount to deposit: "))
        balance = deposit(balance, history, amount)
    elif choice == "3":
        amount = float(input("Enter amount to withdraw: "))
        balance = withdraw(balance, history, amount)
    elif choice == "4":
        show_transaction_history(history)
    elif choice == "5":
        print("Thank you for banking with us. Goodbye!")
        break
    else:
        print("Invalid choice. Please try again.")
```

```python
# Sample run with hardcoded data (no input() needed)

balance = 0.0
history = []

balance = deposit(balance, history, 1000)
balance = withdraw(balance, history, 300)
check_balance(balance)
show_transaction_history(history)
```

<h3 style="font-size:16px; margin-bottom:4px;">3. Inventory Management</h3>

<p style="font-size:13px; margin:4px 0;">Maintain products (name, price, quantity) with functions to add a product, display products, search a product, update quantity, and calculate total inventory value.</p>

```python
def add_product(products, name, price, quantity):
    """Add a new product to the inventory."""
    products.append({"name": name, "price": price, "quantity": quantity})


def display_products(products):
    """Print every product currently in the inventory."""
    if not products:
        print("No products in inventory.")
        return
    for product in products:
        print(f"{product['name']} | Price: {product['price']:.2f} | Quantity: {product['quantity']}")


def search_product(products, name):
    """Return the product dictionary matching the given name, or None."""
    for product in products:
        if product["name"].lower() == name.lower():
            return product
    return None


def update_quantity(products, name, new_quantity):
    """Update the quantity of the named product. Return True if updated."""
    product = search_product(products, name)
    if product is None:
        return False
    product["quantity"] = new_quantity
    return True


def total_inventory_value(products):
    """Return the total value of the inventory (price times quantity, summed)."""
    total_value = 0
    for product in products:
        total_value += product["price"] * product["quantity"]
    return total_value
```

```python
products = []

num_products = int(input("How many products do you want to add? "))
for _ in range(num_products):
    name = input("Product name: ")
    price = float(input("Product price: "))
    quantity = int(input("Product quantity: "))
    add_product(products, name, price, quantity)

print("\nCurrent Inventory:")
display_products(products)

search_name = input("\nEnter a product name to search: ")
found = search_product(products, search_name)
if found:
    print(f"Found: {found}")
else:
    print("Product not found.")

update_name = input("\nEnter a product name to update quantity: ")
new_quantity = int(input("Enter new quantity: "))
if update_quantity(products, update_name, new_quantity):
    print("Quantity updated.")
else:
    print("Product not found.")

print(f"\nTotal Inventory Value: {total_inventory_value(products):.2f}")
```

```python
# Sample run with hardcoded data (no input() needed)

products = []
add_product(products, "Pen", 10.0, 50)
add_product(products, "Notebook", 50.0, 20)

print("Current Inventory:")
display_products(products)

found = search_product(products, "Pen")
print(f"\nFound: {found}")

update_quantity(products, "Notebook", 30)
print("Quantity updated.")

print(f"\nTotal Inventory Value: {total_inventory_value(products):.2f}")
```

<p style="font-size:13px; margin:4px 0;">At least 5 Python questions, displayed one at a time. Accepts answers, checks them, keeps score, and shows the final percentage.</p>

```python
quiz_questions = [
    {"question": "Which keyword is used to define a function in Python?", "answer": "def"},
    {"question": "What data type does len() return?", "answer": "int"},
    {"question": "Which loop is guaranteed to run at least once in Python? (for/while/none)", "answer": "none"},
    {"question": "What symbol starts a comment in Python?", "answer": "#"},
    {"question": "Which built-in function converts a string to an integer?", "answer": "int"},
    {"question": "Which keyword immediately stops a loop?", "answer": "break"},
]


def run_quiz(questions):
    """Ask each question in turn, check the answer, and return the final score."""
    score = 0
    for question_number, question in enumerate(questions, start=1):
        print(f"\nQ{question_number}: {question['question']}")
        user_answer = input("Your answer: ").strip().lower()
        if user_answer == question["answer"].lower():
            print("Correct!")
            score += 1
        else:
            print(f"Wrong. Correct answer: {question['answer']}")
    return score
```

<h3 style="font-size:16px; margin-bottom:4px;">5. Number Analysis Tool</h3>

<p style="font-size:13px; margin:4px 0;">A function that accepts a list and returns the largest number, smallest number, total, average, even count, odd count, positive count, and negative count, without using min(), max(), or sum().</p>

```python
def analyze_numbers(numbers):
    """Return a dictionary of statistics about the list, without using min, max or sum."""
    largest = numbers[0]
    smallest = numbers[0]
    total = 0
    even_count = 0
    odd_count = 0
    positive_count = 0
    negative_count = 0

    for number in numbers:
        if number > largest:
            largest = number
        if number < smallest:
            smallest = number

        total += number

        if number % 2 == 0:
            even_count += 1
        else:
            odd_count += 1

        if number > 0:
            positive_count += 1
        elif number < 0:
            negative_count += 1

    average = total / len(numbers)

    return {
        "largest": largest,
        "smallest": smallest,
        "total": total,
        "average": average,
        "even_count": even_count,
        "odd_count": odd_count,
        "positive_count": positive_count,
        "negative_count": negative_count,
    }
```

```python
numbers = [12, -7, 45, 0, 8, -3, 19, -22, 6, 33]
result = analyze_numbers(numbers)

print(f"Numbers        : {numbers}")
print(f"Largest        : {result['largest']}")
print(f"Smallest       : {result['smallest']}")
print(f"Total          : {result['total']}")
print(f"Average        : {result['average']:.2f}")
print(f"Even Count     : {result['even_count']}")
print(f"Odd Count      : {result['odd_count']}")
print(f"Positive Count : {result['positive_count']}")
print(f"Negative Count : {result['negative_count']}")
```

<h3 style="font-size:16px; margin-bottom:4px;">6. Employee Salary Analyzer</h3>

<p style="font-size:13px; margin:4px 0;">Given employee salaries, functions determine total payroll, average salary, highest salary, lowest salary, and employees earning above average.</p>

```python
def total_payroll(employees):
    """Return the sum of all employee salaries."""
    total = 0
    for employee in employees:
        total += employee["salary"]
    return total


def average_salary(employees):
    """Return the average salary across all employees."""
    return total_payroll(employees) / len(employees)


def highest_salary(employees):
    """Return the employee dictionary with the highest salary."""
    highest = employees[0]
    for employee in employees:
        if employee["salary"] > highest["salary"]:
            highest = employee
    return highest


def lowest_salary(employees):
    """Return the employee dictionary with the lowest salary."""
    lowest = employees[0]
    for employee in employees:
        if employee["salary"] < lowest["salary"]:
            lowest = employee
    return lowest


def above_average_employees(employees):
    """Return a list of employees earning more than the average salary."""
    average = average_salary(employees)
    above_average = []
    for employee in employees:
        if employee["salary"] > average:
            above_average.append(employee)
    return above_average
```

```python
employees = [
    {"name": "Riya", "salary": 55000},
    {"name": "Arjun", "salary": 42000},
    {"name": "Meera", "salary": 68000},
    {"name": "Karan", "salary": 39000},
    {"name": "Sana", "salary": 72000},
]

print(f"Total Payroll  : {total_payroll(employees):.2f}")
print(f"Average Salary : {average_salary(employees):.2f}")
print(f"Highest Salary : {highest_salary(employees)}")
print(f"Lowest Salary  : {lowest_salary(employees)}")

print("\nEmployees earning above average:")
for employee in above_average_employees(employees):
    print(f"{employee['name']} - {employee['salary']}")
```

<p style="font-size:13px; margin:4px 0;">A cart where users can add products, remove products, view the cart, calculate the bill, and exit. The application continues until Exit is selected.</p>

```python
def add_to_cart(cart, name, price, quantity):
    """Add an item to the shopping cart."""
    cart.append({"name": name, "price": price, "quantity": quantity})
    print(f"Added {quantity} x {name} to the cart.")


def remove_from_cart(cart, name):
    """Remove an item from the cart by name. Return True if removed."""
    for item in cart:
        if item["name"].lower() == name.lower():
            cart.remove(item)
            return True
    return False


def view_cart(cart):
    """Print every item currently in the cart."""
    if not cart:
        print("Your cart is empty.")
        return
    for item in cart:
        print(f"{item['name']} | Price: {item['price']:.2f} | Quantity: {item['quantity']}")


def calculate_bill(cart):
    """Return the total bill for everything in the cart."""
    total = 0
    for item in cart:
        total += item["price"] * item["quantity"]
    return total
```

<h3 style="font-size:16px; margin-bottom:4px;">8. Password Strength Checker</h3>

<p style="font-size:13px; margin:4px 0;">A function that checks whether a password contains an uppercase letter, a lowercase letter, a number, a special character, and a minimum of 8 characters, then returns a meaningful strength result.</p>

```python
def check_password_strength(password):
    """Check password criteria and return a strength label with details."""
    special_characters = "!@#$%^&*()-_=+[]{};:,.<>/?"

    has_upper = False
    has_lower = False
    has_digit = False
    has_special = False

    for character in password:
        if character.isupper():
            has_upper = True
        elif character.islower():
            has_lower = True
        elif character.isdigit():
            has_digit = True
        elif character in special_characters:
            has_special = True

    has_min_length = len(password) >= 8

    criteria_met = sum([has_upper, has_lower, has_digit, has_special, has_min_length])

    if criteria_met == 5:
        strength = "Very Strong"
    elif criteria_met == 4:
        strength = "Strong"
    elif criteria_met == 3:
        strength = "Moderate"
    else:
        strength = "Weak"

    details = {
        "has_uppercase": has_upper,
        "has_lowercase": has_lower,
        "has_digit": has_digit,
        "has_special_character": has_special,
        "has_min_length": has_min_length,
    }

    return strength, details
```

```python
password = input("Enter a password to check: ")
strength, details = check_password_strength(password)

print(f"\nStrength: {strength}")
for criterion, met in details.items():
    print(f"{criterion}: {met}")
```

```python
# Sample run with hardcoded data (no input() needed)

password = "Passw0rd!"
strength, details = check_password_strength(password)

print(f"Password : {password}")
print(f"Strength: {strength}")
for criterion, met in details.items():
    print(f"{criterion}: {met}")
```

<h3 style="font-size:16px; margin-bottom:4px;">9. Prime Number Analyzer</h3>

<p style="font-size:13px; margin:4px 0;">Given a range, functions find the prime numbers, count them, calculate their sum, and display the largest prime found.</p>

```python
def is_prime(number):
    """Return True if the number is prime."""
    if number < 2:
        return False
    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1
    return True


def find_primes_in_range(start, end):
    """Return a list of all prime numbers between start and end (inclusive)."""
    primes = []
    for number in range(start, end + 1):
        if is_prime(number):
            primes.append(number)
    return primes


def count_primes(primes):
    """Return how many primes are in the list."""
    count = 0
    for prime in primes:
        count += 1
    return count


def sum_of_primes(primes):
    """Return the sum of all primes in the list."""
    total = 0
    for prime in primes:
        total += prime
    return total


def largest_prime(primes):
    """Return the largest prime in the list, or None if the list is empty."""
    if not primes:
        return None
    largest = primes[0]
    for prime in primes:
        if prime > largest:
            largest = prime
    return largest
```

```python
start = int(input("Enter the start of the range: "))
end = int(input("Enter the end of the range: "))

primes = find_primes_in_range(start, end)

print(f"\nPrime numbers between {start} and {end}: {primes}")
print(f"Count of primes : {count_primes(primes)}")
print(f"Sum of primes   : {sum_of_primes(primes)}")
print(f"Largest prime   : {largest_prime(primes)}")
```

```python
def add_expense(expenses, name, amount):
    """Add a new expense entry."""
    expenses.append({"name": name, "amount": amount})


def view_expenses(expenses):
    """Print every recorded expense."""
    if not expenses:
        print("No expenses recorded yet.")
        return
    for expense in expenses:
        print(f"{expense['name']} : {expense['amount']:.2f}")


def total_expenses(expenses):
    """Return the sum of all recorded expenses."""
    total = 0
    for expense in expenses:
        total += expense["amount"]
    return total


def highest_expense(expenses):
    """Return the expense entry with the highest amount."""
    highest = expenses[0]
    for expense in expenses:
        if expense["amount"] > highest["amount"]:
            highest = expense
    return highest
```

<h3 style="font-size:16px; margin-bottom:4px;">11. Mini Authentication System</h3>

<p style="font-size:13px; margin:4px 0;">A predefined username and password, a maximum number of login attempts, successful and failed login handling, logout, and retry logic.</p>

```python
CORRECT_USERNAME = "admin"
CORRECT_PASSWORD = "super30"
MAX_ATTEMPTS = 3


def login(username, password):
    """Return True if the credentials match the predefined username and password."""
    return username == CORRECT_USERNAME and password == CORRECT_PASSWORD


def logout():
    """Print a logout confirmation message."""
    print("You have been logged out.")
```

```python
logged_in = False
attempts = 0

while attempts < MAX_ATTEMPTS and not logged_in:
    username = input("Username: ")
    password = input("Password: ")

    if login(username, password):
        print("Login successful!")
        logged_in = True
    else:
        attempts += 1
        remaining = MAX_ATTEMPTS - attempts
        print(f"Login failed. Attempts remaining: {remaining}")

if logged_in:
    input("Press Enter to logout...")
    logout()
else:
    print("Maximum login attempts exceeded. Account locked.")
```

<h3 style="font-size:16px; margin-bottom:4px;">12. Super30 Python Utility Application</h3>

<p style="font-size:13px; margin:4px 0;">A menu-driven application with at least five utilities: calculator, palindrome checker, prime checker, factorial calculator, and multiplication table.</p>

```python
def calculator(num1, operator, num2):
    """Perform a basic arithmetic operation and return the result."""
    if operator == "+":
        return num1 + num2
    elif operator == "-":
        return num1 - num2
    elif operator == "*":
        return num1 * num2
    elif operator == "/":
        if num2 == 0:
            return "Error: Division by zero"
        return num1 / num2
    else:
        return "Error: Invalid operator"


def is_palindrome(text):
    """Return True if the text reads the same forwards and backwards."""
    cleaned = text.replace(" ", "").lower()
    return cleaned == cleaned[::-1]


def is_prime_number(number):
    """Return True if the number is prime."""
    if number < 2:
        return False
    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1
    return True


def factorial(number):
    """Return the factorial of a non-negative integer."""
    result = 1
    for value in range(2, number + 1):
        result *= value
    return result


def multiplication_table(number, up_to=10):
    """Print the multiplication table for a number up to a given limit."""
    for multiplier in range(1, up_to + 1):
        print(f"{number} x {multiplier} = {number * multiplier}")
```

```python
while True:
    print("\n----- Super30 Python Utility Application -----")
    print("1. Calculator")
    print("2. Palindrome Checker")
    print("3. Prime Checker")
    print("4. Factorial Calculator")
    print("5. Multiplication Table")
    print("6. Exit")
    choice = input("Enter your choice (1-6): ")

    if choice == "1":
        num1 = float(input("Enter first number: "))
        operator = input("Enter operator (+, -, *, /): ")
        num2 = float(input("Enter second number: "))
        print(f"Result: {calculator(num1, operator, num2)}")
    elif choice == "2":
        text = input("Enter text to check: ")
        if is_palindrome(text):
            print(f"{text} is a palindrome.")
        else:
            print(f"{text} is not a palindrome.")
    elif choice == "3":
        number = int(input("Enter a number: "))
        if is_prime_number(number):
            print(f"{number} is prime.")
        else:
            print(f"{number} is not prime.")
    elif choice == "4":
        number = int(input("Enter a number: "))
        print(f"Factorial of {number} is {factorial(number)}")
    elif choice == "5":
        number = int(input("Enter a number: "))
        multiplication_table(number)
    elif choice == "6":
        print("Goodbye!")
        break
    else:
        print("Invalid choice. Please try again.")
```
