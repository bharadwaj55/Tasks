<h3 style="font-size:16px; margin-bottom:4px;">Task 1: Student Management System</h3>

<p style="font-size:13px; margin:4px 0;">Build a small Student Management System using Python OOP.</p>

<ul style="font-size:13px; margin:4px 0;">
<li>Create a <code>Student</code> class with attributes such as name, email, student ID, course, and marks.</li>
<li>Create multiple student objects and methods to display details, update marks, and calculate average marks.</li>
<li>Use a class variable to track the total number of students and a class method such as <code>get_total_students()</code>.</li>
</ul>

<p style="font-size:13px; margin:4px 0;">Concepts used: Class, objects, <code>__init__</code>, instance methods, class variable, class method.</p>

```python
class Student:
    """Represents a single student and their marks."""

    # Class variable shared by every instance, tracks how many students exist.
    total_students = 0

    def __init__(self, name, email, student_id, course, marks):
        """Initialize a student with their personal and academic details."""
        self.name = name
        self.email = email
        self.student_id = student_id
        self.course = course
        self.marks = marks  # list of marks/scores for the student

        Student.total_students += 1

    def display_details(self):
        """Print a formatted summary of the student's details."""
        print(f"Student ID : {self.student_id}")
        print(f"Name       : {self.name}")
        print(f"Email      : {self.email}")
        print(f"Course     : {self.course}")
        print(f"Marks      : {self.marks}")
        print(f"Average    : {self.calculate_average():.2f}")

    def update_marks(self, new_marks):
        """Replace the student's marks with a new list of marks."""
        self.marks = new_marks

    def add_mark(self, mark):
        """Add a single new mark to the student's existing marks."""
        self.marks.append(mark)

    def calculate_average(self):
        """Return the average of the student's marks, or 0 if there are none."""
        if not self.marks:
            return 0
        return sum(self.marks) / len(self.marks)

    @classmethod
    def get_total_students(cls):
        """Return the total number of Student objects created so far."""
        return cls.total_students
```

```python
# classmethod vs staticmethod
# - @classmethod takes 'cls' as its first argument, so it can read/modify class-level state (e.g. cls.total_students).
# - @staticmethod takes no implicit first argument, so it cannot access class or instance state directly.

# - @classmethod correctly follows subclasses since 'cls' is the class it was actually called on.
# - @staticmethod is used for utility logic that's related to the class but doesn't need class/instance data.

# - Use when: @classmethod for reading/modifying class-level data or factory methods; 
# @staticmethod for utility logic related to the class but independent of class/instance state.

# - get_total_students() needs cls.total_students, so @classmethod is the right choice here.
```

<h4 style="font-size:14px; margin-bottom:2px;">Create multiple student objects</h4>

```python
student1 = Student("Riya", "riya@example.com", "S101", "Data Science", [85, 90, 78])
student2 = Student("Arjun", "arjun@example.com", "S102", "Python Development", [70, 65, 80])
student3 = Student("Meera", "meera@example.com", "S103", "Web Development", [92, 88, 95])

print(f"Total students created: {Student.get_total_students()}")
```

<h4 style="font-size:14px; margin-bottom:2px;">Display student details</h4>

```python
for student in (student1, student2, student3):
    student.display_details()
    print("-" * 30)
```

<h4 style="font-size:14px; margin-bottom:2px;">Update marks and recalculate average</h4>

```python
print(f"Before update, {student2.name}'s average: {student2.calculate_average():.2f}")

student2.update_marks([88, 91, 84])
print(f"After update, {student2.name}'s average: {student2.calculate_average():.2f}")

student1.add_mark(100)
print(f"After adding a mark, {student1.name}'s marks: {student1.marks}")
print(f"After adding a mark, {student1.name}'s average: {student1.calculate_average():.2f}")
```

<h4 style="font-size:14px; margin-bottom:2px;">Track total students with a class method</h4>

```python
student4 = Student("Karan", "karan@example.com", "S104", "Data Science", [60, 75, 70])

print(f"Total students now: {Student.get_total_students()}")
print(f"Same value accessed via instance: {student4.get_total_students()}")
```
