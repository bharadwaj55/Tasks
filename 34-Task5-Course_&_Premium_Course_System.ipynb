<h3 style="font-size:16px; margin-bottom:4px;">Task 5: Course & Premium Course System</h3>

<p style="font-size:13px; margin:4px 0;">Build a simple online learning platform using inheritance: <code>Course</code> is the parent class, <code>PremiumCourse</code> is the child class.</p>

<ul style="font-size:13px; margin:4px 0;">
<li><code>Course</code> contains course name, instructor, duration, and price, with <code>show_course_details()</code> and <code>calculate_discount()</code>.</li>
<li><code>PremiumCourse</code> inherits from <code>Course</code> and adds mentor support and live sessions.</li>
<li>A class variable counts how many courses have been created, with a class method to return the count.</li>
</ul>

<h4 style="font-size:14px; margin-bottom:2px;">Parent class: Course</h4>

```python
class Course:
    """Base class representing a course on the learning platform."""

    # Class variable shared by every course, tracks how many courses exist.
    total_courses = 0

    def __init__(self, course_name, instructor, duration, price):
        """Initialize a course with name, instructor, duration (in hours), and price."""
        self.course_name = course_name
        self.instructor = instructor
        self.duration = duration
        self.price = price

        Course.total_courses += 1

    def show_course_details(self):
        """Print a formatted summary of the course's details."""
        print(f"Course Name : {self.course_name}")
        print(f"Instructor  : {self.instructor}")
        print(f"Duration    : {self.duration} hours")
        print(f"Price       : {self.price}")

    def calculate_discount(self, percentage=10):
        """Return the course price after applying the given discount percentage."""
        return self.price - (self.price * percentage / 100)

    @classmethod
    def get_course_count(cls):
        """Return the total number of Course/PremiumCourse objects created so far."""
        return cls.total_courses
```

<h4 style="font-size:14px; margin-bottom:2px;">Child class: PremiumCourse (inherits from Course)</h4>

```python
class PremiumCourse(Course):
    """Child class that extends Course with mentor support and live sessions."""

    def __init__(self, course_name, instructor, duration, price, mentor_support, live_sessions):
        """Initialize a premium course using the parent Course constructor, plus extra fields."""
        # Reuse the parent class's __init__ so total_courses is also incremented there.
        super().__init__(course_name, instructor, duration, price)
        self.mentor_support = mentor_support
        self.live_sessions = live_sessions

    def show_course_details(self):
        """Print the course details from the parent class, plus premium-specific details."""
        # Call the parent class's method to reuse its display logic.
        super().show_course_details()
        print(f"Mentor Support : {self.mentor_support}")
        print(f"Live Sessions  : {self.live_sessions}")
```

<h4 style="font-size:14px; margin-bottom:2px;">Create Course and PremiumCourse objects</h4>

```python
course1 = Course("Python Basics", "Anil Sharma", 20, 1500)
course2 = Course("SQL Fundamentals", "Neha Gupta", 15, 1200)
premium1 = PremiumCourse("Data Science Bootcamp", "Ravi Kumar", 60, 8000, "1-on-1 mentor calls", "Weekly live sessions")
premium2 = PremiumCourse("Full Stack Web Development", "Priya Nair", 80, 9500, "24x7 mentor chat support", "Daily live sessions")

for course in (course1, course2, premium1, premium2):
    course.show_course_details()
    print("-" * 30)
```

<h4 style="font-size:14px; margin-bottom:2px;">Calculate discount</h4>

```python
print(f"{course1.course_name} price after default 10% discount: {course1.calculate_discount()}")
print(f"{premium1.course_name} price after a 20% discount: {premium1.calculate_discount(20)}")
```

<h4 style="font-size:14px; margin-bottom:2px;">Track total courses with a class method</h4>

```python
print(f"Total courses created: {Course.get_course_count()}")
print(f"Same value accessed via a PremiumCourse instance: {premium2.get_course_count()}")
```
