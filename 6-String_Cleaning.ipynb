```python
message = " Welcome To Python Programming Class "
```

```python
# Remove extra spaces
# strip() returns a copy of the string with leading and trailing whitespace removed
message.strip()
```

```python
# Convert everything to lowercase
# lower() returns a copy of the string converted to lowercase.
message.lower()
```

```python
# Convert everything to uppercase
# upper() returns a copy of the string converted to uppercase
message.upper()
```

```python
# Convert to title case
# Returns a capitalized version of the string
# Since first character is space, it does nothing
message.capitalize()
```

```python
# Replace "Python" with "Advanced Python"
# replace() returns a copy with all occurrences of substring old replaced by new
message.replace("Python","Advanced Python")
```

```python
message = " Welcome To Python Programming Class "
```

```python
# Check whether the string starts with "Welcome"
# startswith() returns True if the string starts with the specified prefix, False otherwise.
# Result is False, because of the leading space
message.startswith("Welcome")
```

```python
# Check whether it ends with "Class"
# endswith() returns True if the string ends with the specified suffix, False otherwise.
# False, because of the trailing space
message.endswith("Class")
```

```python
# count() returns the number of occurrences of given substring in string
# Count occurrences of "o"
message.count("o")
```

```python
# Find the position of "Programming"
# Return -1 on failure.
# Return the lowest index in S where substring sub is found, such that sub is contained within S[start:end]
message.find("Programming")
```

```python
# Find the position of "Programming"
# Return the lowest index in S where substring sub is found, such that sub is contained within S[start:end].
# Raises ValueError when the substring is not found.
message.index("Programming")
```

```python
# Split the sentence into words
# split() return a list of the substrings in the string, using sep as the separator string.
# By default, space is the separator
message.split()
```
