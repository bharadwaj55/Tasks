```python
# Print numbers from 1 to 100 using while.
# Initialized: number = 1 | Condition: number <= 100 | Update/Termination: number += 1 each pass, loop ends once number exceeds 100.

number = 1
while number <= 100:
    print(number, end=" ")
    number += 1
```

```python
# Print numbers from 100 to 1.
# Initialized: number = 100 | Condition: number >= 1 | Update/Termination: number -= 1 each pass, loop ends once number drops below 1.

number = 100
while number >= 1:
    print(number, end=" ")
    number -= 1
```

```python
# Print all even numbers between 1 and 100.
# Initialized: number = 1 | Condition: number <= 100 | Update/Termination: number += 1 each pass, loop ends once number exceeds 100.

number = 1
while number <= 100:
    if number % 2 == 0:
        print(number, end=" ")
    number += 1
```

```python
# Calculate the sum of digits of a number.
# Example: Input: 5832 Output: 18
# Initialized: num = 5832 | Condition: num > 0 | Update/Termination: num = num // 10 strips the last digit each pass, loop ends once num reaches 0.

num = 5832
digit_sum = 0

while num > 0:
    digit = num % 10
    digit_sum = digit_sum + digit
    num = num // 10

print(digit_sum)
```

```python
# Reverse an integer using a while loop.
# Example: Input: 12345 Output: 54321
# Initialized: num = 12345 | Condition: num > 0 | Update/Termination: num = num // 10 strips the last digit each pass, loop ends once num reaches 0.

num = 12345
reversed_num = 0

while num > 0:
    digit = num % 10
    reversed_num = reversed_num * 10 + digit
    num = num // 10

print(reversed_num)
```

```python
# Count the number of digits in an integer.
# Initialized: num = 5832 | Condition: num > 0 | Update/Termination: num = num // 10 strips the last digit each pass, loop ends once num reaches 0.

num = 5832
count = 0

while num > 0:
    count = count + 1
    num = num // 10

print(count)
```

```python
# Calculate factorial using while.
# Initialized: n = 5 | Condition: n > 1 | Update/Termination: n -= 1 each pass, loop ends once n drops to 1.

n = 5
factorial = 1

while n > 1:
    factorial = factorial * n
    n -= 1

print(factorial)
```

```python
# Create a program that repeatedly asks for numbers and stops only when the user enters 0.
# Display the sum of all previously entered numbers.
# Initialized: total = 0 | Condition: while True (infinite) | Termination: break fires as soon as the entered number equals 0.

total = 0

while True:
    number = int(input("Enter a number (0 to stop): "))
    if number == 0:
        break
    total = total + number

print(f"Sum of all entered numbers: {total}")
```

```python
# Create a password checker that keeps asking for the password until the correct password is entered.
# Initialized: password = first user input | Condition: password != correct_password | Termination: loop ends once the entered password matches, updating password on each failed attempt.

correct_password = "python123"
password = input("Enter the password: ")

while password != correct_password:
    print("Incorrect password. Try again.")
    password = input("Enter the password: ")

print("Correct password. Access granted.")
```

```python
# Create a guessing game where the secret number is predefined and the user keeps guessing until correct.
# Initialized: guess = first user input | Condition: guess != secret_number | Termination: loop ends once guess matches, updating guess on each incorrect attempt.

secret_number = 7
guess = int(input("Guess the secret number: "))

while guess != secret_number:
    if guess < secret_number:
        print("Too low. Try again.")
    else:
        print("Too high. Try again.")
    guess = int(input("Guess the secret number: "))

print("Correct! You guessed the secret number.")
```

```python
# Menu-driven calculator using a while loop.
# Initialized: none needed before the loop | Condition: while True (infinite) | Termination: break fires only when the user selects option 5 (Exit).

while True:
    print("\nCalculator Menu")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    if choice == "5":
        print("Exiting calculator.")
        break

    if choice not in ("1", "2", "3", "4"):
        print("Invalid choice. Please try again.")
        continue

    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    if choice == "1":
        print(f"Result: {num1 + num2}")
    elif choice == "2":
        print(f"Result: {num1 - num2}")
    elif choice == "3":
        print(f"Result: {num1 * num2}")
    elif choice == "4":
        if num2 == 0:
            print("Error: Division by zero is not allowed.")
        else:
            print(f"Result: {num1 / num2}")
```

```python
# ATM menu using a while loop.
# Initialized: balance = 1000 before the loop begins | Condition: while True (infinite) | Termination: break fires only when the user selects option 4 (Exit).

balance = 1000

while True:
    print("\nATM Menu")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    choice = input("Enter your choice (1-4): ")

    if choice == "1":
        print(f"Your balance is: {balance}")
    elif choice == "2":
        amount = float(input("Enter amount to deposit: "))
        balance += amount
        print(f"Deposit successful. New balance: {balance}")
    elif choice == "3":
        amount = float(input("Enter amount to withdraw: "))
        if amount > balance:
            print("Insufficient balance.")
        else:
            balance -= amount
            print(f"Withdrawal successful. New balance: {balance}")
    elif choice == "4":
        print("Thank you for using the ATM. Goodbye.")
        break
    else:
        print("Invalid choice. Please try again.")
```
