```python
# Print numbers from 1 to 100 using while.

number = 1
while number <= 100:
    print(number, end=" ")
    number += 1
```

```python
# Print numbers from 100 to 1.

number = 100
while number >= 1:
    print(number, end=" ")
    number -= 1
```

```python
# Print all even numbers between 1 and 100.

number = 1
while number <= 100:
    if number % 2 == 0:
        print(number, end=" ")
    number += 1
```

```python
# Calculate the sum of digits of a number.
# Example: Input: 5832 Output: 18

num = 5832
digit_sum = 0

while num > 0:
    digit = num % 10
    digit_sum = digit_sum + digit
    num = num // 10

print(digit_sum)
```

```python
# Reverse an integer using a while loop.
# Example: Input: 12345 Output: 54321

num = 12345
reversed_num = 0

while num > 0:
    digit = num % 10
    reversed_num = reversed_num * 10 + digit
    num = num // 10

print(reversed_num)
```

```python
# Count the number of digits in an integer.

num = 5832
count = 0

while num > 0:
    count = count + 1
    num = num // 10

print(count)
```

```python
# Calculate factorial using while.

n = 5
factorial = 1

while n > 1:
    factorial = factorial * n
    n -= 1

print(factorial)
```

```python
# Create a program that repeatedly asks for numbers and stops only when the user enters 0.
# Display the sum of all previously entered numbers.

total = 0

while True:
    number = int(input("Enter a number (0 to stop): "))
    if number == 0:
        break
    total = total + number

print(f"Sum of all entered numbers: {total}")
```

```python
# Create a password checker that keeps asking for the password until the correct password is entered.

correct_password = "python123"
password = input("Enter the password: ")

while password != correct_password:
    print("Incorrect password. Try again.")
    password = input("Enter the password: ")

print("Correct password. Access granted.")
```

```python
# Create a guessing game where the secret number is predefined and the user keeps guessing until correct.

secret_number = 7
guess = int(input("Guess the secret number: "))

while guess != secret_number:
    if guess < secret_number:
        print("Too low. Try again.")
    else:
        print("Too high. Try again.")
    guess = int(input("Guess the secret number: "))

print("Correct! You guessed the secret number.")
```
