```python
# Reverse the cart
# option 1 - reverse the string using reverse()
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]
cart.reverse()
cart
```

```python
# option 2 - reverse the string using slicing
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]
reversed_list = cart[::-1]
print(reversed_list)
```

```python
# Sort the products alphabetically
# Option 1: Using sort() (modifies list in place)
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]
cart.sort()
cart
```

```python
# Sort the products alphabetically
# Option 2: Using sorted()
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]
sorted_cart = sorted(cart)
sorted_cart
```

```python
# Explain the concepts

# append() inserts the data passed at the last index
# extend() appends the data at the last index, but the data should be iterable
# insert() is used to insert the data at specific index.
# remove() removes specific data that is passed to the method.
# pop() removes the last data element
# clear() removes all items
# copy() returns a shallow copy of the list
# sort() Sort the list in ascending order and return None. sort is in-place (Modifies original list).
# reverse() reverses the list by Modifying the original list.
```
