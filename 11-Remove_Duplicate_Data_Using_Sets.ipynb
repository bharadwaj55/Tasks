```python
# Students Example
# List of student names with duplicates
students = ["Alice", "Bob", "Charlie", "Alice", "Diana", "Bob", "Eve", "Charlie", "Frank"]
students
```

```python
# Convert the students list to set
unique_students_set=set(students)
unique_students_set
```

```python
# convert unique_students set to list
unique_students_list = list(unique_students_set)
unique_students_list
```

```python
# Print the count of students in original list, unique_students_list

print(f"Count of students in Original list is:", len(students))
print(f"Count of students in Unique students list is:", len(unique_students_list))
print(f"Duplicates removed are: {len(students) - len(unique_students_list)}")
```
