<h3 style="font-size:16px; margin-bottom:4px;">Task 2: Bank Account System</h3>

<p style="font-size:13px; margin:4px 0;">Create a simplified banking application.</p>

<ul style="font-size:13px; margin:4px 0;">
<li>Create a <code>BankAccount</code> class containing account holder name, account number, and balance.</li>
<li>Implement <code>deposit()</code>, <code>withdraw()</code>, <code>check_balance()</code>, and <code>display_account_details()</code>.</li>
<li>Use a class variable for the bank name and a class method that can change the bank name for every account.</li>
<li>Add validation so that a user cannot withdraw more money than available.</li>
<li>Bonus: Keep track of the total number of bank accounts created.</li>
</ul>

```python
class BankAccount:
    """Represents a simple bank account with deposit/withdraw operations."""

    # Class variable shared by every account, holds the bank's name.
    bank_name = "Global Trust Bank"

    # Class variable that tracks how many accounts have been created.
    total_accounts = 0

    def __init__(self, account_holder, account_number, balance=0):
        """Initialize a bank account with holder name, account number, and starting balance."""
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance

        BankAccount.total_accounts += 1

    def deposit(self, amount):
        """Add the given amount to the account balance."""
        if amount <= 0:
            print("Deposit amount must be positive.")
            return
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        """Withdraw the given amount if sufficient balance is available."""
        if amount <= 0:
            print("Withdrawal amount must be positive.")
            return
        if amount > self.balance:
            print(f"Withdrawal failed: insufficient balance. Available: {self.balance}")
            return
        self.balance -= amount
        print(f"Withdrew {amount}. New balance: {self.balance}")

    def check_balance(self):
        """Return the current account balance."""
        return self.balance

    def display_account_details(self):
        """Print a formatted summary of the account's details."""
        print(f"Bank Name      : {BankAccount.bank_name}")
        print(f"Account Holder : {self.account_holder}")
        print(f"Account Number : {self.account_number}")
        print(f"Balance        : {self.balance}")

    @classmethod
    def change_bank_name(cls, new_name):
        """Change the bank name for every account, since it is a shared class variable."""
        cls.bank_name = new_name

    @classmethod
    def get_total_accounts(cls):
        """Return the total number of BankAccount objects created so far."""
        return cls.total_accounts
```

<h4 style="font-size:14px; margin-bottom:2px;">Create multiple bank accounts</h4>

```python
account1 = BankAccount("Riya", "AC1001", 5000)
account2 = BankAccount("Arjun", "AC1002", 2000)
account3 = BankAccount("Meera", "AC1003")

print(f"Total accounts created: {BankAccount.get_total_accounts()}")
```

<h4 style="font-size:14px; margin-bottom:2px;">Display account details</h4>

```python
for account in (account1, account2, account3):
    account.display_account_details()
    print("-" * 30)
```

<h4 style="font-size:14px; margin-bottom:2px;">Deposit and withdraw, including the insufficient-balance check</h4>

```python
account1.deposit(1500)
account1.withdraw(2000)

# Attempt to withdraw more than the available balance.
account2.withdraw(10000)

print(f"{account1.account_holder}'s balance via check_balance(): {account1.check_balance()}")
```

<h4 style="font-size:14px; margin-bottom:2px;">Change the bank name for every account</h4>

```python
print(f"Bank name before change: {account1.bank_name}")

BankAccount.change_bank_name("New Horizon Bank")

print(f"Bank name after change (account1): {account1.bank_name}")
print(f"Bank name after change (account2): {account2.bank_name}")
print(f"Bank name after change (account3): {account3.bank_name}")
```

<h4 style="font-size:14px; margin-bottom:2px;">Bonus: track total number of accounts</h4>

```python
account4 = BankAccount("Karan", "AC1004", 750)

print(f"Total accounts now: {BankAccount.get_total_accounts()}")
print(f"Same value accessed via instance: {account4.get_total_accounts()}")
```
