```python
# calculate total transaction value without using sum().

sum = 0
transactions = [1200, 450, 800, 1500, 2300, 700, 100]
for i in transactions:
    sum = sum + i
print(sum)
```

```python
# From the same list, find the highest transaction without max().

transactions = [1200, 450, 800, 1500, 2300, 700, 100]
n = 0

for i in transactions:
    if i > n:
        n = i
print(f"Highest transaction is {n}")
```

```python
# From the same list, find the lowest transaction without min().

transactions = [1200, 450, 800, 1500, 2300, 700, 100]
n = transactions[0]

for i in transactions:
    if i < n:
        n = i
print(f"Lowest transaction is {n}")
```

```python
# count how many students scored
# 90+
# 75–89
# 50–74
# below 50

marks = [78, 92, 45, 67, 88, 53, 99]
count_90_plus = 0
count_75_89 = 0
count_50_74 = 0
count_below_50 = 0

for mark in marks:
    if mark >= 90:
        count_90_plus = count_90_plus + 1
    elif 75 <= mark <= 89:
        count_75_89 = count_75_89 + 1        
    elif 50 <= mark <= 74:
        count_50_74 = count_50_74 + 1
    else:
        count_below_50 = count_below_50 + 1

print(f"count of Students who scored 90+: {count_90_plus}")
print(f"count of Students who scored between 75 and 89: {count_75_89}")
print(f"count of Students who scored between 50 and 74: {count_50_74}")
print(f"count of Students who scored below 50: {count_below_50}")
```

```python
# Create a simple login system with a maximum of 3 password attempts.

correct_password = "python123"

for attempt in range(1,4):
    password = input(f"Attempt {attempt}: Enter the password")
    if password == correct_password:
        print("Login successful")
        break
    else:
        print("Incorrect password")
else:
    print("Account locked. Maximum attempts exceeded.")
```

```python
# Count the frequency of every character in a string without using Counter.
# Example:
# banana
# b -> 1
# a -> 3
# n -> 2

string = "banana"
frequency = {}

for char in string:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

for char, count in frequency.items():
    print(f"{char} -> {count}")
```

```python
# Find the second-largest number in a list without using sort().

numbers = [55, 62, 19, 81, 108, 729, 543]

largest = numbers[0]
second_largest = numbers[0]

for i in numbers:
    if i > largest:
        second_largest = largest
        largest = i
    elif i > second_largest and i != largest:
        second_largest = i

print(f"Largest number is {largest}")
print(f"Second largest number is: {second_largest}")
```

```python
# Check whether a string is a palindrome using loops.

string = input("Enter the string:")
reversed_string = ""

for char in string:
    reversed_string = char + reversed_string

if string == reversed_string:
    print(f"{string} is a palindrome")
else:
    print(f"{string} is not a palindrome")
```

```python
# Create a basic ATM simulation where a user can repeatedly
# Check balance
# Deposit money
# Withdraw money
# Exit

def atm(balance):
    while True:
        print("\n===== ATM MENU =====")
        print("1. Check Balance")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. Exit")

        choice = input("Enter your choice (1-4): ")

        if choice == "1":
            print(f"Your current balance is: ₹{balance}")

        elif choice == "2":
            amount = float(input("Enter amount to deposit: "))
            if amount > 0:
                balance += amount
                print(f"₹{amount} deposited successfully. New balance: ₹{balance}")
            else:
                print("Invalid amount. Please enter a positive value.")

        elif choice == "3":
            amount = float(input("Enter amount to withdraw: "))
            if amount > balance:
                print("Insufficient balance!")
            elif amount <= 0:
                print("Invalid amount. Please enter a positive value.")
            else:
                balance -= amount
                print(f"₹{amount} withdrawn successfully. New balance: ₹{balance}")

        elif choice == "4":
            print("Thank you for using the ATM. Goodbye!")
            break  # Stop the loop (exit)

        else:
            print("Invalid choice. Please select 1, 2, 3, or 4.")


# Start the ATM with an initial balance
atm(5000)
```
