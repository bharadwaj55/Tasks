```python
# typecasting str to int
# int("100") converts the string "100" into the integer 100
int("100")
```

```python
# typecasting str to float
# float("45.67") converts the string "45.67" into the float 45.67
float("45.67")
```

```python
# typecasting int to str
# str(500) converts the integer 500 into the string '500'
str(500)
```

```python
# typecasting int to bool
# bool(1) converts the integer 1 into the boolean True, since any non-zero number is truthy
bool(1)
```

```python
# typecasting tuple to list
# list((1, 2, 3)) converts the tuple (1, 2, 3) into the list [1, 2, 3]
list((1, 2, 3))
```

```python
# typecasting list to tuple
# tuple([1, 2, 3]) converts the list [1, 2, 3] into the tuple (1, 2, 3)
tuple([1, 2, 3])
```

```python
# typecasting list to set
# set([1, 2, 2, 3]) converts the list [1, 2, 2, 3] into the set {1, 2, 3}, removing the duplicate value
set([1, 2, 2, 3])
```
