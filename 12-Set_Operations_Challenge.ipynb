```python
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}
java_students = {"Priya", "Karan", "Rohit", "Simran"}
```

```python
# Print both sets
print(python_students)
print(java_students)
```

```python
# Find students learning either Python or Java
# use union operation
either_python_or_java = python_students.union(java_students)
print(f"Student learning either Python or Java:",either_python_or_java)
```

```python
# Find students learning both
# use intersection
# Intersection returns set with elements that are common in both datasets.
both = python_students.intersection(java_students)
print(f"Students learning both:",both)
```

```python
# Find students learning only Python
# use difference
# Return a new set with elements in the set that are not in the others.
only_python = python_students.difference(java_students)
print(only_python)
```

```python
# Find students learning only Java
only_java = java_students.difference(python_students)
print(only_java)
```

```python
# Find students belonging to exactly one group
# Symmetric difference includes elements in either set 
# but excludes elements that appear in both sets.
exactly_one = python_students.symmetric_difference(java_students)
print(exactly_one)
```

```python
# Add a new student
# add() Adds an element to a set.
python_students.add("Raj")
python_students
```

```python
# Remove a student
# remove() Remove an element from a set; it must be a member.
# If the element is not a member, raise a KeyError.
python_students.remove("Raj")
python_students
```

```python
# If the element is not a member, raise a KeyError.
python_students.remove("Raj")
python_students
```

```python
# Demonstrate discard()
python_students.discard("Rahul")
python_students
```

```python
# Remove an element from a set if it is a member.
# Unlike set.remove(), the discard() method does not raise an exception 
# when an element is missing from the set.

python_students.discard("Rahul")
python_students
```
