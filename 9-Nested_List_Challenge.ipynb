```python
students = [
    ["Rahul", 21, "Python"],
    ["Priya", 22, "Data Science"],
    ["Aman", 20, "Machine Learning"]
]
```

```python
print(students)
```

```python
# Print the complete list
print(students)
```

```python
# Print Rahul's name
print(students[0][0])
```

```python
# Print Priya's age
print(students[1][1])
```

```python
# Print Aman's course
print(students[2][2])
```

```python
# Print the complete record of Priya
print(students[1])
```

```python
# Change Rahul's course to "AI"
students[0][2]="AI"
```

```python
print(students)
```

```python
# Add another student record manually
students.append(['Raj', 25, 'DS'],)
```

```python
print(students)
```

```python
# Print the updated nested list
print(students)
```

```python
# Explain how indexing works with nested lists.
print(students[1])
```

```python
print(students[2])
```

```python
print(students[3])
```

```python
# Explain how indexing works with nested lists.
# Indexing with Nested Lists works in layers
# you access outer elements first, then inner elements.
# For example, students[1][1] means take the second student record first (students[1], Priya's record),
# then the second item inside it (index 1), which gives Priya's age, 22.
```
