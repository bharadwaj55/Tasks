```python
student = {
"name": "Rahul",
"age": 22,
"course": "Python",
"city": "Bangalore",
"marks": 88
}
```

```python
# Print the complete dictionary
print(student)
```

```python
# Print the student's name
student['name']
```

```python
# Print their course
student['course']
```

```python
# Print all keys
# keys() returns a view of all the keys in the dictionary
print(student.keys())
```

```python
# Print all values
# values() returns a view of all the values in the dictionary
print(student.values())
```

```python
# Print all key-value pairs
# items() returns a view of all the key-value pairs in the dictionary, as (key, value) tuples
print(student.items())
```

```python
# Change marks from 88 to 92
student['marks'] = 92
student
```

```python
# Add "email"
student['email'] = "abc@email.com"
student
```

```python
# Add "phone"
student['phone'] = "123456789"
student
```

```python
# Use update() to modify/add multiple key-value pairs at once
# update() merges the given key-value pairs into the dictionary, adding new keys and overwriting existing ones
# "city": "Mumbai" and "grade": "A" are added here just for the sake of demonstrating update(), not part of the original required task
student.update({"city": "Mumbai", "grade": "A"})
student
```

```python
# Remove "city"
# pop() removes the given key and returns its value; raises a KeyError if the key is not found
student.pop("city")
student
```

```python
# Use get() to retrieve "name"
# get() returns the value for the given key, or None (or a default) instead of raising an error if the key is missing
student.get("name")
```

```python
# Create a copy of the dictionary
# copy() returns a shallow copy of the dictionary, so changes to the copy don't affect the original
student_copy = student.copy()
student_copy
```
