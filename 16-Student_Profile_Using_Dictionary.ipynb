```python
student = {
"name": "Rahul",
"age": 22,
"course": "Python",
"city": "Bangalore",
"marks": 88
}
```

```python
# Print the complete dictionary
print(student)
```

```python
# Print the student's name
student['name']
```

```python
# Print their course
student['course']
```

```python
# Print all keys
print(student.keys())
```

```python
# Print all values
print(student.values())
```

```python
# Print all key-value pairs
student.values()
```

```python
# Change marks from 88 to 92
student['marks'] = "100"
student
```

```python
# Add "email"
student['email'] = "abc@email.com"
student
```

```python
# Add "phone"
student['phone'] = "123456789"
student
```

```python
# Remove "city"
student.pop("city")
student
```

```python
# Use get() to retrieve "name"
student.get("name")
```

```python
# Create a copy of the dictionary
student_copy = student.copy()
student_copy
```
